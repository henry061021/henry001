## 欧几里得算法

实现 gcd 函数求两个整数的最大公约数。

```python
# 1. gcd：欧几里得算法（辗转相除法）
def gcd(m, n):
    # 保证处理过程中余数非负，先取绝对值
    m, n = abs(m), abs(n)
    while n != 0:
        m, n = n, m % n
    return m

# 2. 评估给定数对
print("gcd(2018, 2019) =", gcd(2018, 2019))
print("gcd(1200, 300)  =", gcd(1200, 300))
print("gcd(5040, 60)   =", gcd(5040, 60))
```

### 关于负数参数

在上面的实现中，我们先对两个输入取 `abs()` 再做辗转相除，因此负数输入会被当作其绝对值处理：

- `gcd(-12, 18) = gcd(12, 18) = 6`
- `gcd(-12, -18) = 6`

这种行为是合理的：最大公约数的定义是“同时整除两个数的最大**正**整数”，整除关系对正负号不敏感（若 d 整除 m，则 d 也整除 -m），所以 GCD 在数学上本就应是非负的。

如果**不取绝对值**，纯辗转相除在 Python 中 `%` 的结果符号跟随除数，可能返回负数或 0，例如 `gcd(-12, 18)` 在不处理符号时会得到 `-6` 之类的结果，虽然绝对值相同，但把“最大公约数”定义成负数在数学上不规范。因此取绝对值是更正确的做法。

## 近似欧拉数 e

e = sum_{k=0}^{inf} 1/k!

```python
# 1. euler_limit(n) = (1 + 1/n)^n
def euler_limit(n):
    return (1 + 1 / n) ** n

print("euler_limit(1)      =", euler_limit(1))
print("euler_limit(100)    =", euler_limit(100))
print("euler_limit(1000000)=", euler_limit(1000000))
```

```python
# 2. euler_infinite_sum(n)：公式(1)前 n 项之和（1-indexed）
#    n=0 -> 0；n=1 -> 1（即 k=0 那一项）；n=4 -> 1+1+1/2+1/6
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1  # k!，从 k=0 开始，0! = 1
    for k in range(n):       # k = 0, 1, ..., n-1，共 n 项
        if k == 0:
            fact = 1
        else:
            fact *= k
        total += 1.0 / fact
    return total

print("euler_infinite_sum(0) =", euler_infinite_sum(0))
print("euler_infinite_sum(1) =", euler_infinite_sum(1))
print("euler_infinite_sum(4) =", euler_infinite_sum(4))
print("euler_infinite_sum(20)=", euler_infinite_sum(20))
```

```python
# 3. euler_approx(epsilon)：用 while 循环累加，直到与真实 e 的差在 epsilon 内
import math

def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    k = 0
    fact = 1
    while True:
        if k == 0:
            fact = 1
        else:
            fact *= k
        total += 1.0 / fact
        if abs(total - true_e) < epsilon:
            break
        k += 1
    return total

print("euler_approx(0.01)      =", euler_approx(0.01))
print("euler_approx(1e-6)      =", euler_approx(1e-6))
print("math.exp(1)             =", math.exp(1))
```

```python
# 4. print_euler_sum_table(n)：打印 k=1..n 时 euler_infinite_sum(k) 的值
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

print_euler_sum_table(10)
```

## 测试整数属性

判断一个正整数是否为 2 的幂，再推广到任意底数 b 的幂。

```python
# 1. is_power_of_2(n)：n 是否为 2 的幂（仅用除法和取模，不用 sqrt）
#    思路：不断除以 2，直到不能整除为止；若最终结果为 1 则是 2 的幂。
#    2^0 = 1 也是 2 的幂。
def is_power_of_2(n):
    if n <= 0:
        return False
    while n % 2 == 0:
        n = n // 2
    return n == 1

print("is_power_of_2(1)   =", is_power_of_2(1))    # 2^0 -> True
print("is_power_of_2(2)   =", is_power_of_2(2))    # True
print("is_power_of_2(16)  =", is_power_of_2(16))   # True
print("is_power_of_2(18)  =", is_power_of_2(18))   # False
print("is_power_of_2(1024) =", is_power_of_2(1024)) # True
```

```python
# 2. is_power(b, n)：n 是否为 b 的幂（b, n 均为正整数）
#    思路：不断除以 b，直到不能整除为止；若最终结果为 1 则 n 是 b 的幂。
#    注意 b=1 的特殊情况：只有 n=1 时才是 1 的幂。
def is_power(b, n):
    if b <= 0 or n <= 0:
        return False
    if b == 1:
        return n == 1
    while n % b == 0:
        n = n // b
    return n == 1

print("is_power(2, 1)   =", is_power(2, 1))    # True
print("is_power(2, 16)  =", is_power(2, 16))   # True
print("is_power(3, 27)  =", is_power(3, 27))   # True
print("is_power(3, 28)  =", is_power(3, 28))   # False
print("is_power(10, 1000)=", is_power(10, 1000))# True
print("is_power(1, 1)    =", is_power(1, 1))    # True
print("is_power(1, 2)    =", is_power(1, 2))    # False
```
