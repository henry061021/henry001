## 实验1：基本运算、数据类型与字符串

**思路**：先直观计算四种运算符并打印 `type()`，建立“值—类型—含义”的对应；再通过 `type()` 认识四种基本类型；用 `int()/float()` 做显式转换并捕获非法字符串异常；最后在同一个字符串上演示 strip / 大小写 / 切片 / split / join，并验证字符串不可变、只能通过切片拼接构造新串。

```python
# 1-1 四种运算符：结果、类型与含义
print("8 / 3  =", 8 / 3,  type(8 / 3).__name__,   "# 真除法，结果总是 float")
print("8 // 3 =", 8 // 3, type(8 // 3).__name__,  "# 整除（向下取整），结果为 int")
print("8 % 3  =", 8 % 3,  type(8 % 3).__name__,   "# 取模/取余数")
print("2 ** 7 =", 2 ** 7, type(2 ** 7).__name__,  "# 幂运算，2 的 7 次方")
```

```python
# 1-2 四种基本数据类型及 type() 检查
int_var   = 2024408029                 # 整数
float_var = 3.14                       # 浮点数
str_var   = "Python Data Analysis"     # 字符串
bool_var  = (8 > 3)                    # 布尔值
for v in (int_var, float_var, str_var, bool_var):
    print(f"值 = {v!r:30} 类型 = {type(v).__name__}")
```

```python
# 1-3 类型转换 + 捕获 int("goat") 异常
print('int("123")   =', int("123"),   "| 合法数字字符串 -> int")
print('float("3.14")=', float("3.14"), "| 合法浮点字符串 -> float")
print('int(3.9)     =', int(3.9),     "| 浮点截断取整（向 0 取整），不是四舍五入")

# int("goat") 会抛出 ValueError，因为 "goat" 不是合法整数字面量
try:
    int("goat")
except ValueError as e:
    print("int('goat') 捕获异常：", type(e).__name__, "->", e)
```

```python
# 1-4 字符串处理：去空白、大小写、切片、拆分与 join
s = " Python Data Analysis "
print("原始      :", repr(s))
print("strip()   :", repr(s.strip()))
print("upper()   :", s.strip().upper())
print("lower()   :", s.strip().lower())
print("切片[0:6] :", s.strip()[0:6])      # 取前 6 个字符
words = s.strip().split()               # 默认按空白拆分
print("split()   :", words)
print("join()    :", "-".join(words))     # 用 '-' 把单词重新连接
```

```python
# 1-5 拼接 vs 重复；字符串不可变性
a, b = "Data", "Analysis"
print("拼接 a + ' ' + b =", a + " " + b)
print("重复 'ab' * 3     =", "ab" * 3)

# 字符串不支持按索引直接修改：s[0] = 'X' 会报 TypeError
immutable = "hello"
try:
    immutable[0] = "H"
except TypeError as e:
    print("直接改字符捕获异常：", e)
# 正确做法：切片 + 拼接构造新字符串
new_str = "H" + immutable[1:]
print("构造新字符串      :", new_str)
```

## 实验2：函数、条件、循环与递归

**思路**：用 `if/elif/else` 实现成绩等级分支，越界分数主动 `raise ValueError`；用 for 与 while 两种循环求 1~100 求和并对照结果；用递归与迭代两种方式写阶乘（约定 0!=1）并在 n=0,1,5,10 对照；用元组一次性返回多个统计量并解包；用 `*args` 收集可变参数求和。

```python
# 2-1 score_level(score) 成绩等级
def score_level(score):
    """90~100 优秀；80~<90 良好；60~<80 及格；0~<60 不及格；其余抛 ValueError。"""
    if not (0 <= score <= 100):
        raise ValueError(f"成绩超出 0~100 范围：{score}")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

for sc in (95, 85, 70, 55, 0, 100):
    print(f"score_level({sc}) = {score_level(sc)}")
try:
    score_level(105)
except ValueError as e:
    print("越界测试捕获：", e)
```

```python
# 2-2 for 与 while 分别求 1~100 的整数和
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print("for 结果  :", total_for)
print("while 结果:", total_while)
print("二者相等  :", total_for == total_while)
```

```python
# 2-3 递归版与迭代版阶乘（约定 0! = 1）
def factorial_recursive(n):
    if n <= 1:
        return 1
    return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result

for n in (0, 1, 5, 10):
    r_rec, r_iter = factorial_recursive(n), factorial_iterative(n)
    print(f"n={n:2d}  递归={r_rec:8d}  迭代={r_iter:8d}  一致={r_rec == r_iter}")
```

```python
# 2-4 summary(values) 返回 (最小值, 最大值, 平均值) 元组
def summary(values):
    """接收非空数值列表，返回 (min, max, avg)。"""
    if len(values) == 0:
        raise ValueError("列表不能为空")
    return (min(values), max(values), sum(values) / len(values))

data = [85, 58, 92, 76, 60, 89]
lo, hi, avg = summary(data)   # 元组解包
print(f"数据 = {data}")
print(f"最小值={lo}  最大值={hi}  平均值={avg:.2f}")
```

```python
# 2-5 sum_all(*args) 可变参数求和
def sum_all(*args):
    return sum(args)

print("无参数   :", sum_all())
print("单个参数 :", sum_all(10))
print("多个参数 :", sum_all(1, 2, 3, 4, 5))
```

## 实验3：列表、字典、元组与基础统计

**思路**：用 `zip()` 把姓名与成绩配对成字典，`enumerate(..., start=1)` 输出从 1 开始的序号；用列表推导式做条件筛选与平方变换；用 `len/sum/max/min` 与 f-string 格式化输出平均分（两位小数）和及格率（百分数）；通过“直接赋值 vs `.copy()`”观察引用与拷贝的区别；最后用元组保存记录验证其不可变，并观察 `zip` 以较短序列为准。

```python
# 实验3 数据准备（模拟成绩，不含真实学生信息）
names  = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]
print(names)
print(scores)
```

```python
# 3-1 zip() 构建姓名->成绩字典，items() 逐项输出
name_to_score = dict(zip(names, scores))
print("字典:", name_to_score)
for name, score in name_to_score.items():
    print(f"  {name} -> {score}")
```

```python
# 3-2 enumerate() 输出序号、姓名、成绩（序号从 1 开始）
for idx, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"  第{idx}名: {name} = {score}")
```

```python
# 3-3 列表推导式：及格学生姓名 + 所有成绩的平方
passed_names = [name for name, score in zip(names, scores) if score >= 60]
squares = [s * s for s in scores]
print("及格学生:", passed_names)
print("成绩平方:", squares)
```

```python
# 3-4 基础统计：人数、平均分、最高分、最低分、及格率
count   = len(scores)
avg     = sum(scores) / count
top     = max(scores)
low     = min(scores)
pass_n  = len([s for s in scores if s >= 60])
pass_rt = pass_n / count
print(f"人数={count}  平均分={avg:.2f}  最高分={top}  最低分={low}")
print(f"及格人数={pass_n}  及格率={pass_rt:.1%}")   # 百分数格式
```

```python
# 3-5 alias = scores 与 copied = scores.copy() 的区别（用单独演示列表）
demo = [1, 2, 3]
alias  = demo           # 直接赋值：引用同一列表
copied = demo.copy()    # 浅拷贝：新列表
alias[0] = 999
print("原 demo      :", demo)
print("alias[0]改动后副本 copied :", copied, "  <- 副本不受影响")
```

```python
# 3-6 元组不可变 + 长度不同序列 zip() 的结果
student_record = ("Alice", 85)
print("学生记录元组:", student_record)
try:
    student_record[1] = 90
except TypeError as e:
    print("元组元素重新赋值捕获异常：", e)

# zip 以较短序列为准
short = [1, 2, 3]
long  = ["a", "b", "c", "d", "e"]
print("zip(长短不同)结果:", list(zip(short, long)), "  <- 只配对 3 对")
```

## 实验4：文件读写、异常处理与模块

**思路**：在 Notebook 同级用 `os.makedirs(exist_ok=True)` 建 `exp01_data/`，用 `with open(..., encoding="utf-8")` 写入并读取 students.txt（含 3 条非法记录）；把单行解析封装为 `parse_student()`，字段数量、空姓名、非法整数、成绩越界都抛 `ValueError`；逐行处理时 try/except 捕获异常、记录行号与原因后继续；统计结果写入 report.txt 并演示 `w`（覆盖）与 `a`（追加）；用 pickle 序列化/反序列化并校验相等；把 `parse_student` 写入 `student_utils.py` 后 import 调用；最后读取不存在文件捕获 `FileNotFoundError`。

```python
# 4-1 工作目录、相对路径与创建数据目录
import os
print("当前工作目录 os.getcwd():", os.getcwd())
data_dir = os.path.join("exp01_data")          # Notebook 同级目录
os.makedirs(data_dir, exist_ok=True)
print("数据目录已就绪:", data_dir)
```

```python
# 4-2 写入 students.txt（UTF-8），再读取并跳过表头
students_txt = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
txt_path = os.path.join(data_dir, "students.txt")
with open(txt_path, "w", encoding="utf-8") as f:
    f.write(students_txt)

with open(txt_path, "r", encoding="utf-8") as f:
    lines = f.read().splitlines()
print("原始行数:", len(lines))
print("表头:", lines[0])
print("数据行示例:", lines[1:4], "...")
```

```python
# 4-3 parse_student(line)：解析一行 "name,score"，非法记录抛 ValueError
def parse_student(line):
    line = str(line).strip()
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误：期望 2 个，实际 {len(parts)} -> {line!r}")
    name, score_text = parts[0].strip(), parts[1].strip()
    if not name:
        raise ValueError(f"姓名为空 -> {line!r}")
    try:
        score = int(score_text)
    except ValueError:
        raise ValueError(f"成绩不是合法整数：{score_text!r}") from None
    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0~100：{score}")
    return {"name": name, "score": score}
```

```python
# 4-4 逐行处理：捕获无效记录，记录行号与原因，继续处理
valid_records = []
invalid_records = []   # (行号, 原始行, 原因)

for lineno, raw in enumerate(lines[1:], start=2):   # 跳过表头，行号从 2 开始
    try:
        valid_records.append(parse_student(raw))
    except ValueError as e:
        invalid_records.append((lineno, raw, str(e)))

print("有效记录数:", len(valid_records))
for r in valid_records:
    print("  ", r)
print("无效记录数:", len(invalid_records))
for rec in invalid_records:
    print("  行", rec[0], repr(rec[1]), "->", rec[2])
```

```python
# 4-5 把统计结果写入 report.txt；演示 w 与 a 模式
report_path = os.path.join(data_dir, "report.txt")
valid_n = len(valid_records)
invalid_n = len(invalid_records)
avg_score = sum(r["score"] for r in valid_records) / valid_n
pass_rate = len([r for r in valid_records if r["score"] >= 60]) / valid_n

# w 模式：覆盖写入（清空原有内容）
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数: {valid_n}\n")
    f.write(f"无效记录数: {invalid_n}\n")
    f.write(f"平均分: {avg_score:.2f}\n")
    f.write(f"及格率: {pass_rate:.1%}\n")

# a 模式：追加写入（在文件末尾追加，不覆盖已有内容）
with open(report_path, "a", encoding="utf-8") as f:
    f.write("---- 以上为 w 模式覆盖写入，本行起为 a 模式追加 ----\n")

with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())
```

```python
# 4-6 pickle 保存与读取，验证恢复数据相等
import pickle
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)
with open(pkl_path, "rb") as f:
    loaded = pickle.load(f)
print("pickle 恢复数据与原数据相等:", loaded == valid_records)
print("恢复内容:", loaded)
```

## 实验5：类、继承与对象复制

**思路**：定义 `Student` 类封装 name/score，实现 `is_passed()` 与 `__str__()`；为实验4的有效记录批量创建对象；派生出 `GraduateStudent`，用 `super().__init__()` 复用父类初始化并新增 `research_topic`、重写 `__str__()`；最后用一个含嵌套列表的字典，分别做直接赋值、`copy.copy()`、`copy.deepcopy()` 后修改内部列表，对比三者的引用共享差异。

```python
# 5-1 Student 类：__init__ / is_passed() / __str__()
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name!r}, score={self.score}, passed={self.is_passed()})"
```

```python
# 5-2 为实验4每条有效记录创建 Student 对象
students = [Student(r["name"], r["score"]) for r in valid_records]
for s in students:
    print(s)
```

```python
# 5-3 GraduateStudent 继承 Student，增加 research_topic
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (f"GraduateStudent(name={self.name!r}, score={self.score}, "
                f"passed={self.is_passed()}, topic={self.research_topic!r})")

g = GraduateStudent("Carol", 92, "Machine Learning")
print(g)
print("is_passed():", g.is_passed())
```

```python
# 5-4 直接赋值 / 浅拷贝 / 深拷贝：修改嵌套列表，分别比较
import copy

def make_origin():
    return {"name": "Alice", "scores": [85, 90]}

# 直接赋值：同一对象
o1 = make_origin(); ref = o1
o1["scores"].append(100)
print("直接赋值-共享内部列表:", ref["scores"])

# 浅拷贝：外层新对象，内层列表仍共享
o2 = make_origin(); shallow = copy.copy(o2)
o2["scores"].append(100)
print("浅拷贝-内层列表被影响:", shallow["scores"])

# 深拷贝：完全独立
o3 = make_origin(); deep = copy.deepcopy(o3)
o3["scores"].append(100)
print("深拷贝-内层列表不受影响:", deep["scores"])
```
