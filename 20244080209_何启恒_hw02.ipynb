## 1. 字符串的乐趣

### 1.1 `is_palindrome`：判断回文

忽略空格与大小写，从前往后和从后往前读一致即为回文。

```python
def is_palindrome(s):
    """若 s 是回文（忽略空格、忽略大小写），返回 True，否则返回 False。"""
    # 去掉所有空格并统一转小写
    cleaned = s.replace(' ', '').lower()
    return cleaned == cleaned[::-1]


# 自测
print(is_palindrome("level"))                 # True
print(is_palindrome("kayak"))                 # True
print(is_palindrome("pop"))                   # True
print(is_palindrome("tacocat"))               # True
print(is_palindrome("TACO cat"))               # True
print(is_palindrome("rats live on no evil star"))  # True
print(is_palindrome("Was it a car or a cat I saw"))  # True
print(is_palindrome("hello"))                 # False
```

### 1.2 `is_abecedarian`：字母是否按字典序排列

忽略空格，允许重复字母；逐对比较，后一个字母不小于前一个即为有序。

```python
def is_abecedarian(s):
    """若 s 中字母（忽略空格）按非降序排列，返回 True，否则返回 False。"""
    letters = [ch for ch in s if ch != ' ']
    # 相邻位置两两比较，只要出现逆序即为 False
    for i in range(1, len(letters)):
        if letters[i] < letters[i - 1]:
            return False
    return True


# 自测
print(is_abecedarian("adder"))        # True
print(is_abecedarian("beet"))        # True
print(is_abecedarian("abcd efgh xyz"))  # True
print(is_abecedarian("dog"))         # False
print(is_abecedarian("cat"))         # False
```

### 1.3 `remove_vowels`：去掉所有元音字母

元音为 a、e、i、o、u（大小写均去除）。

```python
def remove_vowels(s):
    """返回去掉所有元音（a/e/i/o/u，不区分大小写）后的字符串。"""
    vowels = set("aeiouAEIOU")
    return ''.join(ch for ch in s if ch not in vowels)


# 自测
print(remove_vowels('cat'))        # ct
print(remove_vowels('audio'))      # d
print(remove_vowels('AEIOUaeiou'))  # 空字符串
print(remove_vowels('hello world')) # hll wrld
```

## 2. 列表的乐趣

### 2.1 `list_reverse`：返回反转后的列表

输入不是列表时引发 `TypeError`。

```python
def list_reverse(lst):
    """返回 lst 的反转副本；lst 不是列表时引发 TypeError。"""
    if not isinstance(lst, list):
        raise TypeError("参数必须是列表（list）")
    return lst[::-1]


# 自测
print(list_reverse([1, 2, 3]))   # [3, 2, 1]
print(list_reverse(['a', 'b']))  # ['b', 'a']
try:
    list_reverse((1, 2, 3))
except TypeError as e:
    print('TypeError:', e)
```

### 2.2 `is_sorted`：判断序列是否按升序排列

只遍历序列一次（一次相邻比较扫描），无需多次遍历；元素需支持比较。

```python
def is_sorted(seq):
    """若 seq 按非降序排列返回 True，否则返回 False；仅遍历一次。"""
    for i in range(1, len(seq)):
        if seq[i] < seq[i - 1]:
            return False
    return True


# 自测
print(is_sorted([1, 2, 3, 4]))      # True
print(is_sorted([1, 2, 2, 3]))      # True
print(is_sorted([3, 1, 2]))         # False
print(is_sorted((1, 5, 9)))         # True（元组也可）
print(is_sorted('abc'))             # True（字符串也可）
```

## 3. 作为向量的元组

### 3.1 `vec_scalar_mult`：向量数乘

校验元组元素与标量类型，类型不合适时引发 `TypeError`。

```python
def vec_scalar_mult(t, s):
    """元组向量 t 乘以标量 s，返回等长元组。"""
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组（tuple）")
    if not _is_real_number(s):
        raise TypeError("标量必须是整数或浮点数")
    for idx, x in enumerate(t):
        if not _is_real_number(x):
            raise TypeError(f"向量第 {idx} 个元素不是整数或浮点数")
    return tuple(x * s for x in t)


# 自测
print(vec_scalar_mult((1, 2, 3), 2))      # (2, 4, 6)
print(vec_scalar_mult((1.0, -2.0), 1.5))  # (1.5, -3.0)
try:
    vec_scalar_mult([1, 2], 2)
except TypeError as e:
    print('TypeError:', e)
```

### 3.2 `vec_inner_product`：向量内积

任一输入不是元组引发 `TypeError`；长度不一致引发 `ValueError`；返回浮点数。

```python
def vec_inner_product(x, y):
    """返回两个元组向量的内积（浮点数）。"""
    if not isinstance(x, tuple):
        raise TypeError("第一个参数必须是元组（tuple）")
    if not isinstance(y, tuple):
        raise TypeError("第二个参数必须是元组（tuple）")
    if len(x) != len(y):
        raise ValueError("两个向量的长度必须一致")
    return float(sum(a * b for a, b in zip(x, y)))


# 自测
print(vec_inner_product((1, 2, 3), (4, 5, 6)))   # 32.0
print(vec_inner_product((1.0, 2.0), (3.0, 4.0)))  # 11.0
try:
    vec_inner_product((1, 2), (1, 2, 3))
except ValueError as e:
    print('ValueError:', e)
```

### 3.3 `check_valid_mx`：校验矩阵是否有效

有效矩阵要求：外层为元组；每个元素都是元组；所有内层元组等长；每个元素都是整数或浮点数。

```python
def check_valid_mx(mx):
    """判断 mx 是否为有效的元组嵌套矩阵。"""
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True
    if not all(isinstance(row, tuple) for row in mx):
        return False
    row_len = len(mx[0])
    if not all(len(row) == row_len for row in mx):
        return False
    for row in mx:
        if not all(_is_real_number(x) for x in row):
            return False
    return True


# 自测
valid = ((1, 2, 3), (4, 5, 6))
print(check_valid_mx(valid))                       # True
print(check_valid_mx(((1, 2), (3,))))             # False（长度不等）
print(check_valid_mx(((1, 'a'), (3, 4))))         # False（含字符串）
print(check_valid_mx([(1, 2), (3, 4)]))            # False（外层是列表）
```

### 3.4 `mx_vec_mult`：矩阵乘向量

先用 `check_valid_mx` 校验矩阵；再校验向量类型与维度，不合理输入引发相应错误。返回结果元组。

```python
def mx_vec_mult(mx, v):
    """矩阵 mx（元组的元组）乘以向量 v（元组），返回结果元组。"""
    if not check_valid_mx(mx):
        raise TypeError("矩阵不是有效的元组嵌套矩阵")
    if not isinstance(v, tuple):
        raise TypeError("向量必须是元组（tuple）")
    if not all(_is_real_number(x) for x in v):
        raise TypeError("向量元素必须是整数或浮点数")
    if len(mx) == 0:
        return ()
    n = len(mx[0])
    if len(v) != n:
        raise ValueError(f"向量维度 {len(v)} 与矩阵列数 {n} 不一致")
    return tuple(
        sum(row[j] * v[j] for j in range(n))
        for row in mx
    )


# 自测：[[1,2],[3,4]] * (5, 6) = (17, 39)
mx = ((1, 2), (3, 4))
print(mx_vec_mult(mx, (5, 6)))   # (17, 39)
print(mx_vec_mult(((1, 0), (0, 1)), (2, 3)))  # (2, 3)
try:
    mx_vec_mult(mx, (1,))
except ValueError as e:
    print('ValueError:', e)
```

## 4. 更多向量的乐趣（面向对象）

### 4.1 / 4.2 `Vector` 类：属性选择与初始化

**维度与元素应作为实例属性，而不是类属性。**

理由：类属性被所有实例共享，而每个向量的维度和元素值都可能不同；若写成类属性，修改一个实例的坐标会影响其他实例，也无法在创建时传入不同数据。因此维度（`dimension`）和元素（`elements`）都应在 `__init__` 中作为实例属性保存。

初始化时：第一个参数为维度（非负整数），第二个参数为元素列表或元组；只给维度不给元素时，默认生成全零向量。维度类型错误、为负数，或与元素个数不一致时抛出异常。

```python
class Vector:
    """用维度 + 元素序列表示的向量。"""

    def __init__(self, dimension, elements=None):
        # 校验维度：必须是整数且非负（布尔值不算整数）
        if isinstance(dimension, bool) or not isinstance(dimension, int):
            raise TypeError("维度必须是整数（int）")
        if dimension < 0:
            raise ValueError("维度不能为负数")

        if elements is None:
            # 只给维度：默认全零向量
            elements = [0] * dimension
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表（list）或元组（tuple）")
        if len(elements) != dimension:
            raise ValueError(f"元素个数 {len(elements)} 与维度 {dimension} 不一致")
        for idx, x in enumerate(elements):
            if not _is_real_number(x):
                raise TypeError(f"第 {idx} 个元素不是整数或浮点数")

        self.dimension = dimension          # 实例属性：维度
        self.elements = tuple(elements)      # 实例属性：元素（统一存为元组）

    # 便于打印查看
    def __repr__(self):
        return f"Vector({self.dimension}, {self.elements!r})"
```

### 4.3 比较运算符：相等与字典序大小比较

坐标相同即相等；否则按元组字典序比较。维度不同时抛出异常。

```python
def _check_same_dim(self, other):
    if not isinstance(other, Vector):
        raise TypeError("比较对象必须是 Vector 实例")
    if self.dimension != other.dimension:
        raise ValueError("只能比较维度相同的两个向量")


# 把方法挂到 Vector 类上
Vector._check_same_dim = _check_same_dim


def __eq__(self, other):
    if not isinstance(other, Vector):
        return NotImplemented
    if self.dimension != other.dimension:
        raise ValueError("只能比较维度相同的两个向量")
    return self.elements == other.elements


def __lt__(self, other):
    self._check_same_dim(other)
    return self.elements < other.elements


def __le__(self, other):
    self._check_same_dim(other)
    return self.elements <= other.elements


def __gt__(self, other):
    self._check_same_dim(other)
    return self.elements > other.elements


def __ge__(self, other):
    self._check_same_dim(other)
    return self.elements >= other.elements


Vector.__eq__ = __eq__
Vector.__lt__ = __lt__
Vector.__le__ = __le__
Vector.__gt__ = __gt__
Vector.__ge__ = __ge__

# 自测
v1 = Vector(2, (2, 4))
v2 = Vector(2, (2, 5))
v3 = Vector(2, (2, 4))
print(v1 == v3)      # True
print(v1 < v2)       # True
print(v1 <= v3)      # True
print(v2 > v1)       # True
print(v1 >= v3)      # True
```

### 4.4 `Vector.dot`：向量内积

参数类型不正确或维度不一致时抛出异常。

```python
def dot(self, other):
    """返回当前向量与另一个 Vector 的内积。"""
    if not isinstance(other, Vector):
        raise TypeError("参数必须是 Vector 实例")
    if self.dimension != other.dimension:
        raise ValueError("两个向量的维度必须一致")
    return sum(a * b for a, b in zip(self.elements, other.elements))


Vector.dot = dot

# 自测：(1, 2)·(3, 4) = 11
print(Vector(2, (1, 2)).dot(Vector(2, (3, 4))))   # 11
```

### 4.5 乘法运算符 `__mul__` / `__rmul__`

- 标量左乘或右乘（`2 * v`、`v * 2`）：各元素乘以标量，返回新 `Vector`；
- 两个向量相乘（`v * w`）：对应位置元素逐一相乘，返回新 `Vector`；
- 两个向量维度不一致时抛出异常。

```python
def _is_scalar(x):
    return _is_real_number(x)


def __mul__(self, other):
    """支持 Vector * 标量 与 Vector * Vector（逐元素相乘）。"""
    if _is_scalar(other):
        return Vector(self.dimension, tuple(x * other for x in self.elements))
    if isinstance(other, Vector):
        if self.dimension != other.dimension:
            raise ValueError("两个向量的维度必须一致")
        return Vector(
            self.dimension,
            tuple(a * b for a, b in zip(self.elements, other.elements)),
        )
    return NotImplemented


def __rmul__(self, other):
    """支持 标量 * Vector；数乘满足交换律，直接复用 __mul__。"""
    if _is_scalar(other):
        return self * other
    return NotImplemented


Vector.__mul__ = __mul__
Vector.__rmul__ = __rmul__

# 自测
v = Vector(2, (2, 3))
w = Vector(2, (4, 5))
print(v * 2)          # Vector(2, (4, 6))
print(2 * v)          # Vector(2, (4, 6))
print(v * w)          # Vector(2, (8, 15))
```
