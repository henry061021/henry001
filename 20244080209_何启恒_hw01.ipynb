## 2. 欧几里得算法（GCD）

### 2.1 实现 `gcd` 函数

```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

# 测试三组数字对
print(gcd(2018, 2019))   # 1
print(gcd(1200, 300))    # 300
print(gcd(5040, 60))     # 60
```

### 2.2 负数参数的讨论

如果输入负数，辗转相除的取模运算在 Python 中仍能收敛（例如 `gcd(-1200, 300)` 会返回 300，`gcd(-2018, 2019)` 返回 1），只是结果的符号取决于辗转过程中 `m % n` 的符号。

这种行为是有意义的：GCD 本质上是"同时能整除两个数的最大正整数"，所以更严谨的做法是最后对结果取绝对值，保证 GCD 始终为正：

```python
def gcd_abs(m, n):
    m, n = abs(m), abs(n)
    while n != 0:
        m, n = n, m % n
    return m

print(gcd(-2018, 2019))   # 负数输入下的表现
print(gcd(-1200, -300))
print(gcd_abs(-1200, -300))  # 取绝对值后始终为正
```

## 3. 近似欧拉数 e

### 3.1 `euler_limit`：用极限 (1+1/n)^n 近似 e

```python
def euler_limit(n):
    return (1 + 1 / n) ** n

print(euler_limit(1))
print(euler_limit(10))
print(euler_limit(1000000))
```

### 3.2 `euler_infinite_sum`：前 n 项阶乘倒数之和（1-索引）

```python
import math

def euler_infinite_sum(n):
    total = 0.0
    for k in range(n):          # k = 0, 1, ..., n-1
        total += 1 / math.factorial(k)
    return total

print(euler_infinite_sum(0))   # 0
print(euler_infinite_sum(1))   # 1
print(euler_infinite_sum(4))   # 1 + 1 + 1/2 + 1/6
print(euler_infinite_sum(10))
```

### 3.3 `euler_approx`：加项直到误差小于 eps

```python
def euler_approx(eps):
    total = 0.0
    k = 0
    term = 1.0
    while term > eps:
        total += term
        k += 1
        term /= k
    return total

print(euler_approx(1e-6))
print(math.exp(1))
```

### 3.4 `print_euler_sum_table`：打印 k=1..n 的近似值

```python
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

print_euler_sum_table(8)
```

## 4. 测试整数属性

### 4.1 `is_power_of_2`：判断是否为 2 的幂（只用除法和 %）

```python
def is_power_of_2(n):
    while n % 2 == 0:
        n //= 2
    return n == 1

for x in [1, 2, 8, 16, 1024, 3, 10, 100]:
    print(x, is_power_of_2(x))
```

### 4.2 `is_power`：推广到判断 n 是否为 b 的幂

```python
def is_power(b, n):
    if b == 1:
        return n == 1
    while n % b == 0:
        n //= b
    return n == 1

print(is_power(2, 16))    # True
print(is_power(3, 27))    # True
print(is_power(3, 10))    # False
print(is_power(1, 1))     # True
print(is_power(1, 5))     # False
```
