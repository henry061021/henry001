### 实验2：函数、条件、循环与递归

````python
# 实验2-1：score_level(score) 等级判定
def score_level(score):
    """根据分数返回等级。

    90~100 -> 优秀; 80~90 -> 良好; 60~80 -> 及格; 0~60 -> 不及格。
    超出 0~100 范围时抛出 ValueError。
    """
    if not isinstance(score, (int, float)):      # 先保证是数值
        raise ValueError(f"分数必须是数值，收到 {type(score).__name__}")
    if score < 0 or score > 100:                 # 再检查取值范围
        raise ValueError(f"分数必须在 0~100 之间，收到 {score}")

    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"


print("【正常输入】")
for s in [100, 95, 90, 85, 80, 60, 59, 0]:
    print(f"  {s:>3} 分 -> {score_level(s)}")

print("\n【越界输入】")
for bad in [-1, 101]:
    try:
        score_level(bad)
    except ValueError as e:
        print(f"  {bad} -> 抛出 ValueError: {e}")
````

````python
# 实验2-2：for 与 while 分别求 1~100 的整数和
total_for = 0
for i in range(1, 101):
    total_for += i

total_while = 0
i = 1
while i <= 100:
    total_while += i
    i += 1

print("for   循环求和结果:", total_for)
print("while 循环求和结果:", total_while)
print("两种实现结果一致吗:", total_for == total_while)
print("用等差公式校验 n(n+1)/2 = 100*101/2 =", 100 * 101 // 2)
````

````python
# 实验2-3：递归版与迭代版阶乘
def factorial_recursive(n):
    """递归求 n!，约定 0! = 1。"""
    if not isinstance(n, int) or isinstance(n, bool) or n < 0:
        raise ValueError(f"n 必须是非负整数，收到 {n!r}")
    if n == 0:
        return 1                       # 递归出口
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    """迭代求 n!，约定 0! = 1。"""
    if not isinstance(n, int) or isinstance(n, bool) or n < 0:
        raise ValueError(f"n 必须是非负整数，收到 {n!r}")
    result = 1
    for k in range(2, n + 1):
        result *= k
    return result


print(f"{'n':>3}{'递归结果':>16}{'迭代结果':>16}{'是否一致':>10}")
for n in [0, 1, 5, 10]:
    r = factorial_recursive(n)
    it = factorial_iterative(n)
    print(f"{n:>3}{r:>16}{it:>16}{str(r == it):>10}")

print("\n校验 10! = 3628800 ->", factorial_recursive(10) == 3628800)
try:
    factorial_recursive(-1)
except ValueError as e:
    print("factorial_recursive(-1) -> ValueError:", e)
````

````python
# 实验2-4：summary(values) 以元组返回最小值、最大值和平均值
def summary(values):
    """接收非空数值列表，返回 (最小值, 最大值, 平均值) 元组。"""
    if not values:
        raise ValueError("values 不能为空")
    return min(values), max(values), sum(values) / len(values)


data = [85, 58, 92, 76, 60, 89]
result = summary(data)
print("返回的元组:", result, "| 类型:", type(result).__name__)

# 通过元组解包接收结果
low, high, avg = summary(data)
print(f"解包后 -> 最小值={low}, 最大值={high}, 平均值={avg:.2f}")

try:
    summary([])
except ValueError as e:
    print("summary([]) -> ValueError:", e)
````

````python
# 实验2-5：可变参数 sum_all(*args)
def sum_all(*args):
    """返回所有参数之和；无参数时返回 0。"""
    print(f"    本次收到 {len(args)} 个参数: {args}")
    return sum(args)


print("【无参数】     结果 =", sum_all())
print("【单个参数】   结果 =", sum_all(10))
print("【多个参数】   结果 =", sum_all(1, 2, 3, 4, 5))
print("【混合数值】   结果 =", sum_all(1.5, 2.5, -1))
````

**实验4-3**：把 `parse_student()` 写入 Notebook 同级的 `student_utils.py` 模块（用 `%%writefile` 单元格魔法生成文件，随后通过 `import` 调用）。

````python
%%writefile student_utils.py
# -*- coding: utf-8 -*-
"""实验01 配套模块：学生成绩记录解析。

本模块提供 parse_student()，把 students.txt 中的一行文本
解析成 {"name": ..., "score": ...} 字典；无效记录抛出 ValueError。
"""


def parse_student(line):
    """解析一行 "name,score" 文本，返回字典。

    参数:
        line (str): 形如 "Alice,85" 的一行文本。

    返回:
        dict: {"name": str, "score": int}

    异常:
        ValueError: 字段数量不对 / 姓名为空 / 成绩不是整数 / 成绩超出 0~100。
    """
    # 1) 去掉首尾空白
    line = str(line).strip()

    # 2) 按英文逗号切分
    parts = line.split(",")

    # 3) 检查字段数量
    if len(parts) != 2:
        raise ValueError(
            f"字段数量错误：期望 2 个字段(name,score)，实际 {len(parts)} 个 -> {line!r}"
        )

    name, score_text = parts[0].strip(), parts[1].strip()

    # 4) 检查姓名是否为空
    if not name:
        raise ValueError(f"姓名为空 -> {line!r}")

    # 5) 成绩转换为整数
    try:
        score = int(score_text)
    except ValueError:
        raise ValueError(f"成绩不是合法整数：{score_text!r} -> {line!r}") from None

    # 6) 检查成绩范围
    if not 0 <= score <= 100:
        raise ValueError(f"成绩超出 0~100 范围：{score} -> {line!r}")

    return {"name": name, "score": score}
````

````python
# 实验4-4：逐行处理文件，捕获无效记录异常，继续处理后续记录
valid_records = []      # 有效记录（字典列表）
invalid_records = []    # (行号, 原始内容, 错误原因)

with open(data_file, "r", encoding="utf-8") as f:
    lines = f.read().splitlines()

for lineno, line in enumerate(lines, start=1):
    if lineno == 1:            # 跳过表头
        continue
    if not line.strip():       # 跳过空行
        continue
    try:
        valid_records.append(parse_student(line))
    except ValueError as e:
        # 记录出错的行号与原因，然后继续处理后面的记录
        invalid_records.append((lineno, line, str(e)))

print(f"【有效记录】共 {len(valid_records)} 条：")
for r in valid_records:
    print("   ", r)

print(f"\n【无效记录】共 {len(invalid_records)} 条：")
for lineno, line, reason in invalid_records:
    print(f"    第 {lineno} 行 | {line!r} | 原因: {reason}")
````

````python
# 实验4-5：把统计结果写入 report.txt，并演示 w 与 a 模式
valid_count = len(valid_records)
invalid_count = len(invalid_records)
total_scores = [r["score"] for r in valid_records]
avg_score = sum(total_scores) / valid_count
pass_count = sum(1 for s in total_scores if s >= 60)
pass_rate = pass_count / valid_count

report_file = os.path.join(data_dir, "report.txt")
summary_text = (
    f"有效记录数: {valid_count}\n"
    f"无效记录数: {invalid_count}\n"
    f"平均分    : {avg_score:.2f}\n"
    f"及格率    : {pass_rate:.1%}\n"
)

# ---------- w 模式：覆盖写入 ----------
with open(report_file, "w", encoding="utf-8") as f:
    f.write("===== 实验01 统计报告 =====\n")
    f.write(summary_text)
print("[第 1 次 w 模式] 写入完成")

# 再以 w 模式写一次，观察是否覆盖
with open(report_file, "w", encoding="utf-8") as f:
    f.write("===== 实验01 统计报告 (第 2 次 w 写入) =====\n")
    f.write(summary_text)
print("[第 2 次 w 模式] 写入完成")
with open(report_file, "r", encoding="utf-8") as f:
    print("\n---- w 模式后的文件内容 ----")
    print(f.read(), end="")

# ---------- a 模式：追加写入 ----------
with open(report_file, "a", encoding="utf-8") as f:
    f.write("\n[追加行] 以上报告由 exp01 实验生成。\n")
print("---- a 模式后的文件内容 ----")
with open(report_file, "r", encoding="utf-8") as f:
    print(f.read(), end="")

print("\n说明：w 模式会先清空文件原有内容再写入；a 模式保留原有内容，在末尾追加。")
````

### 实验5：类、继承与对象复制

````python
# 实验5-1：定义 Student 类
class Student:
    """学生类：保存姓名和成绩，并提供是否及格判断。"""

    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        """成绩 >= 60 视为及格。"""
        return self.score >= 60

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"Student(name={self.name!r}, score={self.score}, 状态={status})"


demo_student = Student("Alice", 85)
print(demo_student)
print("is_passed() ->", demo_student.is_passed())
print("不及格示例  ->", Student("Bob", 58))
````

````python
# 实验5-2：为实验4中的每条有效记录创建 Student 对象
students = [Student(r["name"], r["score"]) for r in valid_records]
print(f"共创建 {len(students)} 个 Student 对象：")
for s in students:
    print(f"    {s}  | is_passed() = {s.is_passed()}")
````

````python
# 实验5-3：定义继承自 Student 的 GraduateStudent
class GraduateStudent(Student):
    """研究生：在 Student 的基础上增加研究方向。"""

    def __init__(self, name, score, research_topic):
        super().__init__(name, score)      # 用 super() 初始化父类属性
        self.research_topic = research_topic

    def __str__(self):                     # 重写 __str__()
        status = "及格" if self.is_passed() else "不及格"
        return (f"GraduateStudent(name={self.name!r}, score={self.score}, "
                f"研究方向={self.research_topic!r}, 状态={status})")


gs = GraduateStudent("Alice", 92, "机器学习")
print(gs)
print("is_passed() ->", gs.is_passed())
print("是 Student 的实例吗:", isinstance(gs, Student))
print("是 GraduateStudent 的实例吗:", isinstance(gs, GraduateStudent))
````

````python
# 实验5-4：直接赋值 / 浅拷贝 / 深拷贝
import copy

def make_origin():
    """每次都新建一个独立的原始对象，保证三次比较互不干扰。"""
    return {"name": "Alice", "scores": [85, 90]}


# ---------- 1. 直接赋值 ----------
o1 = make_origin()
a1 = o1                      # 同一个对象的另一个名字
a1["scores"].append(100)
print("【1. 直接赋值 alias = origin】")
print("  修改前 origin:", make_origin())
print("  修改后 origin:", o1)
print("  修改后 alias :", a1)
print("  origin is alias ->", o1 is a1, "  => 是同一个对象，改一个另一个也变")
print()

# ---------- 2. 浅拷贝 copy.copy() ----------
o2 = make_origin()
a2 = copy.copy(o2)
a2["scores"].append(100)
print("【2. 浅拷贝 copy.copy()】")
print("  修改后 origin:", o2)
print("  修改后 copy  :", a2)
print("  origin is copy          ->", o2 is a2,
      "  => 外层字典是新对象")
print("  origin['scores'] is copy['scores'] ->", o2["scores"] is a2["scores"],
      "  => 内层列表仍是同一个，所以内层被改了")
print()

# ---------- 3. 深拷贝 copy.deepcopy() ----------
o3 = make_origin()
a3 = copy.deepcopy(o3)
a3["scores"].append(100)
print("【3. 深拷贝 copy.deepcopy()】")
print("  修改后 origin  :", o3)
print("  修改后 deepcopy:", a3)
print("  origin['scores'] is deepcopy['scores'] ->", o3["scores"] is a3["scores"],
      "  => 内层列表也被复制，修改副本完全不影响原对象")
````
